# Fase 1: Definición del proyecto

## Análisis de factores asociados a licitaciones públicas desiertas en Chile

**Curso:** Programación Para la Ciencia de Datos (MCDI500)  
**Fuente:** ChileCompra - Mercado Público (Licitaciones)
**Período:** Primer semestre del 2026 (Enero a junio)

### 1. Contextualización del problema

Las compras públicas permiten que los organismos del Estado
adquieran bienes y servicios necesarios para cumplir sus
funciones, sin embargo, algunas licitaciones terminan declaradas
desiertas, lo que puede dificultar la adquisición oportuna
de los bienes y servicios requeridos.

Este proyecto busca identificar características asociadas
a este fenómeno mediante el análisis de datos públicos
de ChileCompra.

### 2. Preguntas de investigación

¿Qué características de las licitaciones publicadas
durante el primer semestre del 2026, que recibieron al menos una
oferta, se asocian con una mayor probabilidad de terminar
declaradas desiertas?¿Cómo puede utilizarse esta información
para apoyar decisiones preventivas de los organismos compradores para evitar que se declaron desiertas?

### 3. Objetivos del proyecto

#### 3.1. Objetivo general

* Identificar y analizar las características de las licitaciones publicadas en Chile durante el primer semestre del 2026, que recibieron al menos una oferta, asociadas con una mayor probabilidad de ser declaradas desiertas.

#### 3.2. Objetivos específicos

* Identificar patrones y factores asociados a la declaración de licitaciones desiertas mediante técnicas de análisis exploratorio y visualizaciones.

* Proponer recomendaciones basadas en la evidencia obtenida que permitan a los organismos compradores priorizar una revisión preventiva.

#### 4. Justificación técnica

El análisis de las licitaciones públicas requiere procesar un volumen considerable de registros administrativos y identificar características relevantes. Para abordar esta problemática se utilizará Python, junto con bibliotecas como Pandas y NumPy, que permiten realizar operaciones de exploración, preparación, transformación y análisis de datos.

Se utilizará JupyterLab para el desarrollo de notebooks reproducibles, integrando código, documentación y resultados. Git y GitHub permitirán gestionar las versiones del proyecto y registrar los avances del equipo.

#### 5. Alcances del proyecto

* El estudio se limita a los registros de licitaciones públicas de ChileCompra correspondientes al período enero-junio de 2026.

* Se considerarán aquellas licitaciones que recibieron al menos una oferta.

* Se examinarán características económicas, temporales, institucionales y relacionadas con los procedimientos de contratación.

* Para el análisis se considerarán aquellas licitaciones declaradas desiertas.

* El proyecto no pretende demostrar relaciones causales ni explicar todos los casos de licitaciones desiertas del sistema de compras públicas chileno.

### 6. Supuestos y restricciones

#### 6.1. Supuestos

- Los registros provienen de las bases de datos oficiales publicadas por ChileCompra.
- El código de licitación permitirá identificar procesos únicos.
- Las variables disponibles contienen información relevante para caracterizar los procedimientos.
- Los estados registrados podrán utilizarse para construir una variable de resultado (luego de verificar su significado).

#### 6.2. Restricciones

- Los archivos disponibles incluyen registros de licitaciones con ofertas, por lo que no permiten estudiar directamente los procedimientos sin oferentes.
- Una misma licitación puede aparecer en múltiples filas debido a sus ofertas u otros registros asociados.
- Algunos procedimientos presentan estados distintos de adjudicada o desierta.
- Los datos pueden contener valores faltantes, inconsistencias y problemas de formato.
- Los montos pueden encontrarse expresados en diferentes unidades monetarias.
- El período de estudio comprende únicamente enero-junio de 2026.
- Las asociaciones estadísticas identificadas no demostrarán por sí mismas relaciones de causa y efecto.
